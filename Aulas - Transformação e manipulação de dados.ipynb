### AULA 2

```python
colunas = list(dados.columns)
colunas
```

```python
dados = dados.explode(colunas[3:])
dados
```

```python
dados.reset_index(inplace = True, drop = True)
dados.head()
```

```python
dados.info()
```

```python
import numpy as np
# Pq essa biblioteca fornece um tipo de dado que engolba números inteiros e float.
```

```python
dados['max_hospedes'] = dados['max_hospedes'].astype(np.int64)
```

```python
col_intnumericas = ['quantidade_banheiros','quantidade_quartos','quantidade_camas']
```

```python
dados[col_intnumericas] = dados[col_intnumericas].astype(np.int64)
```

```python
dados['avaliacao_geral'] = dados['avaliacao_geral'].astype(np.float64)
```

```python
dados['preco']
```

```python
#removendo o $ utilizando apply e lambda, o strip apaga espaços em branco antes e depois do valor
dados['preco'].apply(lambda x: x.replace('$','').replace(',','').strip())
```

```python
dados['preco'] = dados['preco'].apply(lambda x: x.replace('$','').replace(',','').strip())
```

```python
dados['preco'] = dados['preco'].astype(np.float64)
```

```python
dados.info()
```

```python
# o Applymap é usado para substituir as mesmas coisas em colunas diferentes
dados[['taxa_deposito','taxa_limpeza']] = dados[['taxa_deposito','taxa_limpeza']].applymap(lambda x: x.replace('$','').replace(',','').strip())
```

```python
dados[['taxa_deposito','taxa_limpeza']] = dados[['taxa_deposito','taxa_limpeza']].astype(np.float64)
```

```python
dados.info()
```

```python
dados['descricao_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex = True)
```
